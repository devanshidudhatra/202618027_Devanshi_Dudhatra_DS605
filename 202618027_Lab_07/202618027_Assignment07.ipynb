```python
import csv
import random
import time
from sklearn.linear_model import LinearRegression

# Load data
data = []

with open("garments_worker_productivity.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        data.append(row)

# Features and target
features = [
    "targeted_productivity",
    "smv",
    "wip",
    "over_time",
    "incentive",
    "idle_time",
    "idle_men",
    "no_of_style_change",
    "no_of_workers"
]

target = "actual_productivity"

X = []
Y = []

for row in data:
    x = []
    valid = True

    for feature in features:
        if row[feature] == "":
            valid = False
            break
        x.append(float(row[feature]))

    if row[target] == "":
        valid = False

    if valid:
        X.append(x)
        Y.append(float(row[target]))

# Train-test split
random.seed(42)

indices = list(range(len(X)))
random.shuffle(indices)

split = int(0.8 * len(X))

train_indices = indices[:split]
test_indices = indices[split:]

X_train = [X[i] for i in train_indices]
X_test = [X[i] for i in test_indices]

Y_train = [Y[i] for i in train_indices]
Y_test = [Y[i] for i in test_indices]

# Model
model = LinearRegression()

start = time.perf_counter()
model.fit(X_train, Y_train)
train_time = time.perf_counter() - start

start = time.perf_counter()
predictions = model.predict(X_test)
prediction_time = time.perf_counter() - start

# MAE
mae = 0

for i in range(len(Y_test)):
    mae += abs(Y_test[i] - predictions[i])

mae = mae / len(Y_test)

# RMSE
rmse = 0

for i in range(len(Y_test)):
    rmse += (Y_test[i] - predictions[i]) ** 2

rmse = (rmse / len(Y_test)) ** 0.5

# R2
mean_y = sum(Y_test) / len(Y_test)

ss_res = 0
ss_total = 0

for i in range(len(Y_test)):
    ss_res += (Y_test[i] - predictions[i]) ** 2
    ss_total += (Y_test[i] - mean_y) ** 2

r2 = 1 - (ss_res / ss_total)

print("LIBRARY IMPLEMENTATION")
print("----------------------")
print("Training Time  :", train_time)
print("Prediction Time:", prediction_time)
print("MAE            :", mae)
print("RMSE           :", rmse)
print("R2 Score       :", r2)
```

```python
import csv
import random
import time
from sklearn.linear_model import LogisticRegression

# Load data
data = []

with open("garments_worker_productivity.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        data.append(row)

# Features
features = [
    "targeted_productivity",
    "smv",
    "wip",
    "over_time",
    "incentive",
    "idle_time",
    "idle_men",
    "no_of_style_change",
    "no_of_workers"
]

X = []
Y = []

for row in data:

    x = []
    valid = True

    for feature in features:

        if row[feature] == "":
            valid = False
            break

        x.append(float(row[feature]))

    if row["actual_productivity"] == "":
        valid = False

    if valid:
        X.append(x)

        productivity = float(row["actual_productivity"])

        if productivity >= 0.5:
            Y.append(1)
        else:
            Y.append(0)

# Train-test split
random.seed(42)

indices = list(range(len(X)))
random.shuffle(indices)

split = int(0.8 * len(X))

train_indices = indices[:split]
test_indices = indices[split:]

X_train = [X[i][:] for i in train_indices]
X_test = [X[i][:] for i in test_indices]

Y_train = [Y[i] for i in train_indices]
Y_test = [Y[i] for i in test_indices]

# Standardization
means = []
stds = []

for j in range(len(features)):

    values = [X_train[i][j] for i in range(len(X_train))]

    mean = sum(values) / len(values)

    variance = 0

    for value in values:
        variance += (value - mean) ** 2

    std = (variance / len(values)) ** 0.5

    if std == 0:
        std = 1

    means.append(mean)
    stds.append(std)

for i in range(len(X_train)):
    for j in range(len(features)):
        X_train[i][j] = (X_train[i][j] - means[j]) / stds[j]

for i in range(len(X_test)):
    for j in range(len(features)):
        X_test[i][j] = (X_test[i][j] - means[j]) / stds[j]

# Logistic Regression
model = LogisticRegression(max_iter=1000)

start = time.perf_counter()

model.fit(X_train, Y_train)

train_time = time.perf_counter() - start

# Prediction
start = time.perf_counter()

predictions = model.predict(X_test)

prediction_time = time.perf_counter() - start

# Accuracy
correct = 0

for i in range(len(Y_test)):
    if predictions[i] == Y_test[i]:
        correct += 1

accuracy = correct / len(Y_test)

print("LIBRARY IMPLEMENTATION")
print("----------------------")
print("Training Time  :", train_time)
print("Prediction Time:", prediction_time)
print("Accuracy       :", accuracy)
```

```python
import csv
import random
import time
import math

# Load data
data = []

with open("garments_worker_productivity.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        data.append(row)

# Features
features = [
    "targeted_productivity",
    "smv",
    "wip",
    "over_time",
    "incentive",
    "idle_time",
    "idle_men",
    "no_of_style_change",
    "no_of_workers"
]

X = []
Y = []

for row in data:

    x = []
    valid = True

    for feature in features:

        if row[feature] == "":
            valid = False
            break

        x.append(float(row[feature]))

    if row["actual_productivity"] == "":
        valid = False

    if valid:
        X.append(x)

        productivity = float(row["actual_productivity"])

        if productivity >= 0.5:
            Y.append(1)
        else:
            Y.append(0)

# Train-test split
random.seed(42)

indices = list(range(len(X)))
random.shuffle(indices)

split = int(0.8 * len(X))

train_indices = indices[:split]
test_indices = indices[split:]

X_train = [X[i][:] for i in train_indices]
X_test = [X[i][:] for i in test_indices]

Y_train = [Y[i] for i in train_indices]
Y_test = [Y[i] for i in test_indices]

# Standardization
means = []
stds = []

for j in range(len(features)):

    values = [X_train[i][j] for i in range(len(X_train))]

    mean = sum(values) / len(values)

    variance = 0

    for value in values:
        variance += (value - mean) ** 2

    std = (variance / len(values)) ** 0.5

    if std == 0:
        std = 1

    means.append(mean)
    stds.append(std)

for i in range(len(X_train)):
    for j in range(len(features)):
        X_train[i][j] = (X_train[i][j] - means[j]) / stds[j]

for i in range(len(X_test)):
    for j in range(len(features)):
        X_test[i][j] = (X_test[i][j] - means[j]) / stds[j]

# Weights
b0 = 0
b = [0] * len(features)

learning_rate = 0.01
epochs = 1000

# Gradient Descent
start = time.perf_counter()

for epoch in range(epochs):

    db0 = 0
    db = [0] * len(features)

    for i in range(len(X_train)):

        z = b0

        for j in range(len(features)):
            z += b[j] * X_train[i][j]

        # Sigmoid
        if z >= 0:
            p = 1 / (1 + math.exp(-z))
        else:
            e = math.exp(z)
            p = e / (1 + e)

        error = p - Y_train[i]

        db0 += error

        for j in range(len(features)):
            db[j] += error * X_train[i][j]

    n = len(X_train)

    db0 = db0 / n

    for j in range(len(features)):
        db[j] = db[j] / n

    # Update weights
    b0 = b0 - learning_rate * db0

    for j in range(len(features)):
        b[j] = b[j] - learning_rate * db[j]

train_time = time.perf_counter() - start

# Prediction
start = time.perf_counter()

predictions = []

for i in range(len(X_test)):

    z = b0

    for j in range(len(features)):
        z += b[j] * X_test[i][j]

    if z >= 0:
        p = 1 / (1 + math.exp(-z))
    else:
        e = math.exp(z)
        p = e / (1 + e)

    if p >= 0.5:
        predictions.append(1)
    else:
        predictions.append(0)

prediction_time = time.perf_counter() - start

# Accuracy
correct = 0

for i in range(len(Y_test)):
    if predictions[i] == Y_test[i]:
        correct += 1

accuracy = correct / len(Y_test)

print("FROM SCRATCH IMPLEMENTATION")
print("---------------------------")
print("Training Time  :", train_time)
print("Prediction Time:", prediction_time)
print("Accuracy       :", accuracy)
```
