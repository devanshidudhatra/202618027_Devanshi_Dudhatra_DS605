**Task 2 - Vectorized Arithmetic and Linear Algebra**

● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @
or np.matmul().\
● For a square matrix, find its transpose and determinant.\
● If invertible, find the inverse and verify A @ A^-1 using np.allclose().\
● Use vectorized NumPy operations; avoid explicit Python loops.

```python
A = np.array([
    [7, 11],
    [30, 43]
])

B = np.array([
    [50, 36],
    [27, 18]
])
```

```python
addition = A + B
print("\nA + B:")
print(addition)

# Corresponding elements are multiplied.
element_multiplication = A * B
print("\nA * B (Element-wise multiplication):")
print(element_multiplication)

# @ performs actual matrix multiplication.
matrix_multiplication = A @ B
print("\nA @ B (Matrix multiplication):")
print(matrix_multiplication)
```

```python
transpose = A.T
print("\nTranspose of A:")
print(transpose)

determinant = np.linalg.det(A)
print("\nDeterminant of A:")
print(determinant)
```

```python
# A matrix is invertible when its determinant is not zero.

if not np.isclose(determinant, 0):

    # Find inverse of A
    inverse = np.linalg.inv(A)
    print("\nInverse of A:")
    print(inverse)

    # Verify A multiplied by its inverse
    result = A @ inverse
    print("\nA @ A^-1:")
    print(result)

    # Check whether the result is approximately
    # equal to the identity matrix.
    identity = np.eye(A.shape[0])

    print("\nIs A @ A^-1 equal to identity matrix?")
    print(np.allclose(result, identity))

else:
    print("\nA is not invertible because its determinant is zero.")
```

**Task 8 - Features and Pivot Table**\
● Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.\
● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest
groups.

```python
df = pd.read_csv('D:\\DAU\\Sem-1\\ML\\202618027_Devanshi_Dudhatra_DS605\\202618027_Lab_02\\train.csv')
# 1. Create FamilySize
# FamilySize includes the passenger plus their
# siblings/spouses and parents/children.
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
print("FamilySize:")
print(df[["SibSp", "Parch", "FamilySize"]].head())

# 2. Create IsAlone
# If FamilySize is 1, the passenger is travelling alone.
# Otherwise, the passenger is travelling with family.
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print("\nIsAlone:")
print(df[["FamilySize", "IsAlone"]].head())

# 3. Create Pivot Table
# Rows = Sex
# Columns = Pclass
# Values = mean Survived
pivot = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("\nPivot Table:")
print(pivot)

# 4. Find highest and lowest groups
highest = pivot.stack().idxmax()
highest_value = pivot.stack().max()

lowest = pivot.stack().idxmin()
lowest_value = pivot.stack().min()

print("\nHighest survival group:")
print("Sex:", highest[0])
print("Pclass:", highest[1])
print("Survival Rate:", highest_value)

print("\nLowest survival group:")
print("Sex:", lowest[0])
print("Pclass:", lowest[1])
print("Survival Rate:", lowest_value)
```

**Task 9 - Visualizations and Observations**\
● Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative
relationships.\
● Plot the survival rate by Sex and report the male/female survival rates.\

```python
import seaborn as sns
df = pd.read_csv('D:\\DAU\\Sem-1\\ML\\202618027_Devanshi_Dudhatra_DS605\\202618027_Lab_02\\train.csv')


# --------------------------------------------------
# 1. Correlation Heatmap
# --------------------------------------------------

# Select numerical columns
numerical_data = df[
    ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
]

correlation = numerical_data.corr()

print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# Find strongest positive and negative relationships
# --------------------------------------------------

# Remove correlation of each variable with itself
correlation_without_diagonal = correlation.copy()

for column in correlation_without_diagonal.columns:
    correlation_without_diagonal.loc[column, column] = 0

strongest_positive = correlation_without_diagonal.stack().idxmax()
positive_value = correlation_without_diagonal.stack().max()

strongest_negative = correlation_without_diagonal.stack().idxmin()
negative_value = correlation_without_diagonal.stack().min()

print("\nStrongest Positive Relationship:")
print(strongest_positive, ":", positive_value)

print("\nStrongest Negative Relationship:")
print(strongest_negative, ":", negative_value)


# --------------------------------------------------
# 2. Survival Rate by Sex
# --------------------------------------------------

survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("\nSurvival Rate by Sex:")
print(survival_by_sex)

survival_by_sex.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# --------------------------------------------------
# 3. Age vs Fare by Survival
# --------------------------------------------------

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```
