Step - 1: We need to create a scrapy project 

Use - scrapy startproject tutorial

Why this is needed?
when the above command is ran in terminal a scrapy folder is created and it helps Scrapy to know that where the spider is, what setting are used, etc. If this is not done then scarpy won't be able to find the spider and scarpping could not be done.

What actually is a Spider?
it is like a web crawler. in the spider folder we create a class Spider because it navigates the scarpy that which is the website, which webpage to start and the whole pipeline is written in it.

Step 2: After Scrapy project is created we need to create a Spider

Use - scrapy genspider books books.toscrape.com
the spider named 'books' is created and it is linked to the url provided

Why this?
as given above scrapy uses it as a pipeline like what to do and if spider is not present than scrapy will not know how to proceed.

Why was it hard to do this code in ipynb format?
Because scrapy is not just a library. it is a framework like flask or django which needs it structure to work properly.

Step 3: After creating all that it is time to create Spider

Reference - spiders/books.py

Step 4: Now we need to run the spider to scrape all the data into csv file named raw_books.csv

use - scrapy crawl books

how this command actually works?
it first goes to scrapy.cfg to look where is the setting of the project in which spider is created.
then it goes to settings.py where all the meta data is there
then it looks inside spiders folder to find the 'books' spider
once it finds books.py it runs the code where first class object is created and then crawler is started

Step 5: Once the spider completes its task raw_books.csv is created and data is stored in it

Now it can be used Report the total number of scraped records, missing values, and duplicate UPC values.
