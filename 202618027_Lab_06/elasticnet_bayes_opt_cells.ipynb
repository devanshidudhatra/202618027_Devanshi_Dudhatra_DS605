## Search method: Bayesian Optimization (Gaussian Process surrogate + Expected Improvement)

Not grid search, not random search, not manual search. After every Oracle call the code **learns from all results so far** and *calculates* where to query next:

1. **Start:** a middle configuration + a few points chosen to be as far apart as possible (deterministic, no randomness).
2. **Surrogate model:** a Gaussian Process (written from scratch: kernel, Cholesky, linear solves) predicts, for every untried configuration, the expected loss and how uncertain that prediction is. It learns by itself which hyperparameters matter.
3. **Acquisition (Expected Improvement):** scores each untried configuration by how much better than the current best it is expected to be (low predicted loss + high uncertainty both raise the score).
4. **Query only the top-scored configuration** (1 Oracle call), add the result, refit, repeat. Scoring configurations uses the surrogate only, never the Oracle.
5. **Stop** when expected improvement is ~0, when the best has stalled, or at `BUDGET`.
